## 3. Validación de intervalos de confianza mediante bootstrap
En la etapa 1 se estimaron IC 95 % *t* para la media, que suponen normalidad aproximada del promedio (Teorema Central del Límite). Aquí se validan con **bootstrap no paramétrico**: se remuestrea con reemplazo la muestra observada B = 10.000 veces, se recalcula el estadístico en cada remuestra y la distribución de esas réplicas aproxima la distribución muestral del estimador, sin suponer una forma.

Se comparan dos intervalos bootstrap:
- **Percentil**: percentiles 2,5 y 97,5 de las réplicas.
- **BCa** (*bias-corrected and accelerated*): ajusta los percentiles por el sesgo de las réplicas (*z₀*) y por la asimetría del estimador (aceleración *a*, estimada con jackknife).

Parámetros validados: media de `atraso_semestres`, media de `edad_titulacion` y diferencia de medias de atraso entre hombres y mujeres (IC de Welch de la etapa 1).

```python
B = 10_000                          # número de remuestras (mínimo exigido)
rng = np.random.default_rng(42)     # semilla fija, igual que en la etapa 1

def boot_media(x, B=B, bloque=100):
    """Réplicas bootstrap de la media: B remuestras con reemplazo de tamaño n.
    Se generan de a 100 para no saturar la memoria."""
    n = len(x)
    reps = np.empty(B)
    for i in range(0, B, bloque):
        idx = rng.integers(0, n, size=(min(bloque, B - i), n))
        reps[i:i + bloque] = x[idx].mean(axis=1)
    return reps

def jack_media(x):
    """Medias jackknife: la media dejando fuera cada observación, una a la vez."""
    return (x.sum() - x) / (len(x) - 1)

def ic_percentil(reps, conf=0.95):
    alfa = 1 - conf
    return np.percentile(reps, [100 * alfa / 2, 100 * (1 - alfa / 2)])

def ic_bca(reps, theta, jack, conf=0.95):
    """IC BCa: corrige los percentiles por sesgo (z0) y asimetría (a)."""
    alfa = 1 - conf
    z0 = stats.norm.ppf((reps < theta).mean())           # sesgo de las réplicas
    d = jack.mean() - jack
    a = (d ** 3).sum() / (6 * ((d ** 2).sum()) ** 1.5)   # aceleración (jackknife)
    z = stats.norm.ppf([alfa / 2, 1 - alfa / 2])
    p = stats.norm.cdf(z0 + (z0 + z) / (1 - a * (z0 + z)))
    return np.percentile(reps, 100 * p), z0, a
```

```python
x_atraso = df["atraso_semestres"].dropna().to_numpy()
x_edad   = df["edad_titulacion"].dropna().to_numpy()
h = df.loc[df["sexo"] == "Hombre", "atraso_semestres"].dropna().to_numpy()
m = df.loc[df["sexo"] == "Mujer",  "atraso_semestres"].dropna().to_numpy()

# Diferencia de medias: se remuestrea cada grupo por separado (bootstrap estratificado)
boot = {
    "media atraso": (x_atraso.mean(), boot_media(x_atraso), jack_media(x_atraso)),
    "media edad":   (x_edad.mean(),   boot_media(x_edad),   jack_media(x_edad)),
    "dif. atraso H−M": (h.mean() - m.mean(),
                        boot_media(h) - boot_media(m),
                        np.concatenate([jack_media(h) - m.mean(), h.mean() - jack_media(m)])),
}

ic_ref = {
    "media atraso":    (REF_IC_MEDIA.loc["atraso_semestres", "IC95_inf"], REF_IC_MEDIA.loc["atraso_semestres", "IC95_sup"]),
    "media edad":      (REF_IC_MEDIA.loc["edad_titulacion", "IC95_inf"],  REF_IC_MEDIA.loc["edad_titulacion", "IC95_sup"]),
    "dif. atraso H−M": (REF_WELCH["IC95_inf"], REF_WELCH["IC95_sup"]),
}

filas = []
for nombre, (theta, reps, jack) in boot.items():
    perc = ic_percentil(reps)
    bca, z0, a = ic_bca(reps, theta, jack)
    t_inf, t_sup = ic_ref[nombre]
    filas.append({"parámetro": nombre, "estimado": theta,
                  "IC t etapa 1": f"[{t_inf:.3f}; {t_sup:.3f}]",
                  "IC percentil": f"[{perc[0]:.3f}; {perc[1]:.3f}]",
                  "IC BCa": f"[{bca[0]:.3f}; {bca[1]:.3f}]",
                  "ancho t": t_sup - t_inf, "ancho perc": perc[1] - perc[0], "ancho BCa": bca[1] - bca[0],
                  "EE boot": reps.std(ddof=1), "z0": z0, "a": a})

tabla_boot = pd.DataFrame(filas).set_index("parámetro")
tabla_boot.round(4)
```

```python
fig, ejes = plt.subplots(1, 3, figsize=(13, 3.8))

for ax, (nombre, (theta, reps, jack)) in zip(ejes, boot.items()):
    perc = ic_percentil(reps)
    bca, _, _ = ic_bca(reps, theta, jack)
    t_inf, t_sup = ic_ref[nombre]

    ax.hist(reps, bins=50, color="0.75", edgecolor="white")
    ax.axvline(theta, color="black", linewidth=1.5, label="Estimado")
    for v in (t_inf, t_sup):
        ax.axvline(v, color="black", linestyle=":", label="IC t etapa 1" if v == t_inf else None)
    for v in perc:
        ax.axvline(v, color="tab:blue", linestyle="--", label="IC percentil" if v == perc[0] else None)
    for v in bca:
        ax.axvline(v, color="tab:red", linestyle="-.", label="IC BCa" if v == bca[0] else None)
    ax.set_title(nombre)
    ax.set_xlabel("Valor del estadístico en cada remuestra")

ejes[0].set_ylabel("Frecuencia (B = 10.000)")
ejes[0].legend(fontsize=8)
fig.tight_layout()
fig.savefig(RUTA_FIG / "fig_bootstrap_ic.png", dpi=300)
plt.show()
```

### Análisis de la validación bootstrap
**Concordancia.** Para los tres parámetros los IC percentil y BCa coinciden con los IC clásicos de la etapa 1 hasta el tercer decimal (diferencias de ancho < 0,002). La media de atraso (3,398; IC *t* [3,367; 3,429]) se reproduce exactamente; en edad y en la diferencia H−M los límites difieren a lo más en 0,001. El error estándar bootstrap (0,016; 0,020; 0,032) es prácticamente igual al error estándar analítico de la etapa 1.

**Por qué coinciden pese a la asimetría.** `atraso_semestres` es muy asimétrica (asimetría 4,41 en la etapa 1) y no es normal; sin embargo, el IC *t* no supone normalidad de los datos sino de la **media muestral**. Con n = 91.667 el Teorema Central del Límite se cumple holgadamente: las distribuciones bootstrap son simétricas y acampanadas, la corrección de sesgo *z₀* ≈ 0 y la aceleración *a* ≈ 0,002. Por eso BCa no corrige casi nada respecto del percentil.

**Supuestos y confiabilidad.** El IC *t* supone normalidad aproximada del estimador e independencia; el bootstrap supone que la muestra representa a la población e independencia entre observaciones. La independencia se cumple casi totalmente: solo 461 personas (0,88 % de las filas) tienen más de un título. Con este tamaño muestral los tres métodos son igualmente confiables para la media; BCa sería preferible en muestras pequeñas o con estimadores sesgados, donde *z₀* y *a* se alejan de 0.

**Limitación del bootstrap con datos discretos.** En la etapa 1 el IC bootstrap de la mediana de atraso fue [2; 2]: la variable es entera y casi todas las remuestras tienen la misma mediana. No es un error sino una propiedad de los datos, y es la razón por la que aquí se valida la media y no la mediana.

**Conclusión.** Los IC de la etapa 1 quedan **validados**: son robustos al método de estimación.

## 4. Validación de pruebas de hipótesis mediante permutación
Un test de permutación no supone normalidad ni varianzas iguales: solo supone que, **si H0 es cierta, las etiquetas de grupo son intercambiables**. Se mantiene fijo el estadístico (diferencia entre grupos), se reasignan al azar las etiquetas B = 10.000 veces y se obtiene la distribución del estadístico bajo H0. El valor *p* de permutación es la proporción de permutaciones con un estadístico al menos tan extremo como el observado (bilateral), con la corrección (conteo + 1)/(B + 1).

Se validan las dos pruebas de la etapa 1:
- **Prueba 1** (*t* de Welch): diferencia de medias de atraso entre hombres y mujeres.
- **Prueba 2** (chi-cuadrado 2×2): diferencia en el % de atraso alto (> 4 semestres) entre jornada vespertina y diurna. En una tabla 2×2 el chi-cuadrado equivale a contrastar esta diferencia de proporciones.

```python
def test_permutacion(valores, en_grupo_a, B=B):
    """Test de permutación de dos grupos. Estadístico: media del grupo A − media del grupo B.
    Bajo H0 las etiquetas son intercambiables: se reasignan al azar B veces."""
    n, n_a = len(valores), en_grupo_a.sum()
    obs = valores[en_grupo_a].mean() - valores[~en_grupo_a].mean()
    total = valores.sum()
    dist = np.empty(B)
    for b in range(B):
        s_a = valores[rng.permutation(n)[:n_a]].sum()      # n_a valores al azar forman el "grupo A"
        dist[b] = s_a / n_a - (total - s_a) / (n - n_a)
    p = (np.sum(np.abs(dist) >= abs(obs)) + 1) / (B + 1)  # bilateral, con corrección +1
    return obs, dist, p

# Prueba 1: atraso según sexo
sub1 = df[df["atraso_semestres"].notna()]
obs1, dist1, p1 = test_permutacion(sub1["atraso_semestres"].to_numpy(),
                                   (sub1["sexo"] == "Hombre").to_numpy())

# Prueba 2: % de atraso alto, vespertina vs diurna (en puntos porcentuales)
sub2 = df[df["jornada"].isin(["Diurna", "Vespertina"]) & df["atraso_semestres"].notna()]
obs2, dist2, p2 = test_permutacion(100 * sub2["atraso_alto"].to_numpy(dtype=float),
                                   (sub2["jornada"] == "Vespertina").to_numpy())

tabla_perm = pd.DataFrame({
    "estadístico observado": [obs1, obs2],
    "máx |permutado|":       [np.abs(dist1).max(), np.abs(dist2).max()],
    "obs / DE permutación":  [obs1 / dist1.std(), obs2 / dist2.std()],
    "p paramétrico etapa 1":      [REF_WELCH["p"], REF_CHI2["p"]],
    "p permutación":         [p1, p2],
    "p mínimo posible":      [1 / (B + 1)] * 2,
}, index=["Welch: atraso H − M (semestres)", "Chi²: % atraso alto Vesp − Diur (pp)"])
tabla_perm
```

```python
fig, ejes = plt.subplots(1, 2, figsize=(11, 3.8))
paneles = [
    (ejes[0], dist1, obs1, "Prueba 1: atraso según sexo", "Diferencia de medias H − M (semestres)"),
    (ejes[1], dist2, obs2, "Prueba 2: jornada y atraso alto", "Diferencia de % atraso alto Vesp − Diur (pp)"),
]
for ax, dist, obs, titulo, xlab in paneles:
    ax.hist(dist, bins=50, color="0.75", edgecolor="white", label="Distribución bajo H0")
    ax.axvline(obs, color="tab:red", linewidth=2, label=f"Observado = {obs:.2f}")
    ax.set_title(titulo)
    ax.set_xlabel(xlab)
    ax.legend(fontsize=8)

ejes[0].set_ylabel("Frecuencia (10.000 permutaciones)")
fig.tight_layout()
fig.savefig(RUTA_FIG / "fig_permutacion.png", dpi=300)
plt.show()
```

### Análisis de la validación por permutación
**Concordancia.** Ambas pruebas de la etapa 1 se confirman. En la prueba 1 la diferencia observada (1,068 semestres) es casi nueve veces mayor que la máxima obtenida en 10.000 permutaciones (0,120); en la prueba 2 la diferencia de 9,06 pp supera ampliamente el máximo permutado (1,98 pp). En ningún caso una permutación alcanzó el valor observado, por lo que *p* = 1/(B + 1) ≈ 0,0001 en ambas.

**Por qué los *p* difieren en magnitud sin contradecirse.** El *p* paramétrico de la etapa 1 (1,1·10⁻²³³ y 1,0·10⁻⁷⁰) proviene de una distribución teórica y puede ser arbitrariamente pequeño; el *p* de permutación tiene un piso de 1/(B + 1), así que solo puede informarse como *p* < 0,0001. Ambos llevan a la misma decisión (rechazar H0 con α = 0,05). La concordancia es también cuantitativa: el estadístico observado está a 33,8 desviaciones estándar de la distribución de permutación (*t* de Welch = 32,76) en la prueba 1, y a 17,6 en la prueba 2 (√χ² = √316,09 = 17,8).

**Supuestos y método más apropiado.** Welch supone normalidad aproximada de las medias (garantizada por el TCL con n ≈ 40.000 y 51.000 por grupo) y admite varianzas distintas, lo que es necesario aquí porque Levene las rechazó en la etapa 1. La permutación no supone ninguna distribución, pero su H0 es más estricta (distribuciones idénticas): con varianzas distintas no aísla solo la diferencia de medias. Para este caso, Welch es la prueba principal adecuada y la permutación es una confirmación libre de supuestos distribucionales. Con muestras pequeñas o datos muy asimétricos la permutación ganaría relevancia.

**Significancia no es magnitud.** Con n tan grande cualquier diferencia real resulta significativa. Ambos efectos son pequeños (d de Cohen = 0,23; V de Cramér = 0,06): la conclusión robusta es que **hay** diferencias por sexo y jornada, no que sean grandes.

## 5. Estabilidad de correlaciones mediante bootstrap
Se evalúa si las correlaciones de Spearman reportadas en la etapa 1 son estables frente al remuestreo. Para cada par se generan B = 10.000 remuestras **de filas completas** (se conserva el emparejamiento de las dos variables), se recalcula ρ en cada una y se obtiene su IC 95 % percentil.

Criterio: una correlación es **robusta** si su IC no incluye 0 y es estrecho; es **inestable o nula** si el IC incluye 0 o es ancho. Se analizan cuatro pares: atraso–edad, semestres reales–edad, duración formal–edad y duración formal–atraso. Se excluye semestres reales–atraso (ρ = 0,82) porque el atraso se define como semestres reales − duración formal: esa correlación es por construcción y no aporta información.

Como todas las variables son discretas (semestres enteros y edad en meses), los rangos de cada remuestra se calculan por conteo de valores, lo que da exactamente el mismo ρ que `scipy.stats.spearmanr` en una fracción del tiempo.

```python
def rangos_promedio(codigos, k):
    """Rangos promedio (con empates) a partir del código de cada valor único."""
    conteo = np.bincount(codigos, minlength=k)
    antes = np.cumsum(conteo) - conteo          # cuántos valores son menores
    return (antes + (conteo + 1) / 2)[codigos]

def spearman_boot(x, y, B=B):
    """Réplicas bootstrap de rho de Spearman remuestreando pares (x, y) completos."""
    _, cx = np.unique(x, return_inverse=True)
    _, cy = np.unique(y, return_inverse=True)
    kx, ky, n = cx.max() + 1, cy.max() + 1, len(x)
    reps = np.empty(B)
    for b in range(B):
        i = rng.integers(0, n, n)
        reps[b] = np.corrcoef(rangos_promedio(cx[i], kx), rangos_promedio(cy[i], ky))[0, 1]
    return reps

def fuerza(r):
    r = abs(r)
    return "despreciable" if r < 0.1 else "débil" if r < 0.3 else "moderada" if r < 0.5 else "fuerte"

PARES = [("atraso_semestres", "edad_titulacion"),
         ("semestres_reales", "edad_titulacion"),
         ("dur_total_carr",   "edad_titulacion"),
         ("dur_total_carr",   "atraso_semestres")]

boot_corr, filas = {}, []
for v1, v2 in PARES:
    datos = df[[v1, v2]].dropna()
    x, y = datos[v1].to_numpy(), datos[v2].to_numpy()
    rho = stats.spearmanr(x, y).statistic
    reps = spearman_boot(x, y)
    li, ls = ic_percentil(reps)
    nombre = f"{v1} – {v2}"
    boot_corr[nombre] = reps
    incluye_0 = li <= 0 <= ls
    filas.append({"par": nombre, "n": len(datos), "rho etapa 1": REF_SPEARMAN[(v1, v2)], "rho": rho,
                  "IC95 inf": li, "IC95 sup": ls, "ancho": ls - li, "EE boot": reps.std(ddof=1),
                  "incluye 0": incluye_0,
                  "estabilidad": "nula / inestable" if incluye_0 or (ls - li) > 0.1 else "robusta",
                  "fuerza": fuerza(rho)})

tabla_corr = pd.DataFrame(filas).set_index("par")
tabla_corr.round(4)
```

### Análisis de estabilidad de las correlaciones
**Reproducción y estabilidad.** Las cuatro correlaciones reproducen los valores de la etapa 1 (diferencias < 0,005, atribuibles al redondeo a dos decimales en la etapa 1). Sus distribuciones bootstrap son simétricas y muy concentradas: error estándar ≈ 0,003 e IC 95 % de ancho entre 0,012 y 0,014. Ningún IC incluye el 0, por lo que, según el criterio definido, **las cuatro son estables frente al remuestreo**.

**Estable no significa relevante.** Con n ≈ 92.000–105.000 el bootstrap estima el ρ con gran precisión, de modo que incluso asociaciones mínimas resultan distintas de 0. Por eso se clasifican también por fuerza:
- **atraso–edad (ρ = 0,47; IC [0,460; 0,472])** y **semestres reales–edad (ρ = 0,46; IC [0,454; 0,468])**: moderadas y robustas. Parte de esta relación es esperable: cada semestre adicional de permanencia suma medio año a la edad de titulación.
- **duración formal–edad (ρ = −0,19; IC [−0,196; −0,183])**: débil pero robusta. Las carreras más largas se asocian a titulados algo más jóvenes; una explicación posible (no verificada aquí) es que las carreras cortas concentran más ingreso tardío o programas para trabajadores.
- **duración formal–atraso (ρ = 0,009; IC [0,002; 0,016])**: el IC excluye el 0 por muy poco, pero el valor es **despreciable**. En la práctica el atraso no depende de la duración formal de la carrera: las carreras largas no acumulan más semestres de atraso que las cortas.

**Limitación.** El bootstrap mide la variabilidad muestral, no la heterogeneidad entre subgrupos. Una correlación estable en el total podría cambiar por área o tipo de institución; eso se aborda en el análisis de robustez.

## 6. Simulación Monte Carlo: ¿qué observaría una sola cohorte?
**Escenario.** La etapa 1 detectó que los hombres se atrasan 1,07 semestres más que las mujeres usando 91.667 titulados. Sin embargo, una universidad o carrera que analiza su propia cohorte trabaja con decenas o cientos de casos. Se simula qué observaría esa institución: cuánto varía, solo por azar, el atraso de una cohorte a otra, y con qué probabilidad (potencia) una prueba de Welch detectaría la diferencia por sexo según el tamaño de la cohorte (n = 50, 100, 200, 500 y 1.000).

**Modelo (solo parámetros de la etapa 1).**
- Sexo de cada titulado ~ Bernoulli con P(mujer) = 51.592 / 91.667 = 0,563.
- Atraso según sexo ~ **binomial negativa**, ajustada por método de momentos a la media y desviación estándar de cada grupo en la etapa 1 (hombres: 3,999 y 5,434; mujeres: 2,931 y 4,101). Se elige porque el atraso es un conteo discreto, asimétrico y con varianza mucho mayor que la media (sobredispersión), lo que descarta la normal y la Poisson.
- Atraso alto: más de 4 semestres, igual que en la etapa 1.

Para cada tamaño se simulan R = 10.000 cohortes independientes.

```python
P_MUJER = REF_WELCH["n_m"] / (REF_WELCH["n_h"] + REF_WELCH["n_m"])

def params_nb(media, de):
    """Binomial negativa por método de momentos: media = r(1-p)/p y var = r(1-p)/p²."""
    r = media ** 2 / (de ** 2 - media)
    p = r / (r + media)
    return r, p

NB_H = params_nb(REF_WELCH["media_h"], REF_WELCH["sd_h"])
NB_M = params_nb(REF_WELCH["media_m"], REF_WELCH["sd_m"])

# Comparación modelo vs datos observados (h y m vienen de la sección 3)
chequeo = pd.DataFrame({
    "media modelo":       [stats.nbinom(*NB_H).mean(), stats.nbinom(*NB_M).mean()],
    "media observada":    [h.mean(), m.mean()],
    "DE modelo":          [stats.nbinom(*NB_H).std(), stats.nbinom(*NB_M).std()],
    "DE observada":       [h.std(ddof=1), m.std(ddof=1)],
    "% alto modelo":      [100 * stats.nbinom(*NB_H).sf(UMBRAL_ALTO), 100 * stats.nbinom(*NB_M).sf(UMBRAL_ALTO)],
    "% alto observado":   [100 * (h > UMBRAL_ALTO).mean(), 100 * (m > UMBRAL_ALTO).mean()],
    "% negativos obs.":   [100 * (h < 0).mean(), 100 * (m < 0).mean()],
}, index=["Hombres", "Mujeres"])

print(f"P(mujer) = {P_MUJER:.4f}")
print(f"BN hombres: r = {NB_H[0]:.4f}, p = {NB_H[1]:.4f}")
print(f"BN mujeres: r = {NB_M[0]:.4f}, p = {NB_M[1]:.4f}")
chequeo.round(3)
```

```python
R = 10_000                         # iteraciones Monte Carlo por tamaño
TAMANOS = [50, 100, 200, 500, 1000]
ALFA = 0.05

def simular_cohortes(n, R=R):
    """R cohortes de n titulados: sexo ~ Bernoulli(P_MUJER), atraso ~ BN según sexo (parámetros etapa 1).
    Devuelve, por cohorte: media de atraso, % atraso alto, diferencia H − M y si Welch rechaza H0."""
    mujer = rng.random((R, n)) < P_MUJER
    atraso = np.where(mujer, rng.negative_binomial(*NB_M, size=(R, n)),
                             rng.negative_binomial(*NB_H, size=(R, n))).astype(float)
    n_m = mujer.sum(axis=1)
    n_h = n - n_m
    # Sumas y sumas de cuadrados por grupo, cohorte a cohorte
    s_m = np.where(mujer, atraso, 0).sum(axis=1);       s_h = atraso.sum(axis=1) - s_m
    q_m = np.where(mujer, atraso ** 2, 0).sum(axis=1);  q_h = (atraso ** 2).sum(axis=1) - q_m
    m_m, m_h = s_m / n_m, s_h / n_h
    v_m = (q_m - n_m * m_m ** 2) / (n_m - 1)
    v_h = (q_h - n_h * m_h ** 2) / (n_h - 1)
    # t de Welch con grados de libertad de Welch-Satterthwaite
    ee2_h, ee2_m = v_h / n_h, v_m / n_m
    t = (m_h - m_m) / np.sqrt(ee2_h + ee2_m)
    gl = (ee2_h + ee2_m) ** 2 / (ee2_h ** 2 / (n_h - 1) + ee2_m ** 2 / (n_m - 1))
    p = 2 * stats.t.sf(np.abs(t), gl)
    return pd.DataFrame({"media_atraso": atraso.mean(axis=1),
                         "pct_alto": 100 * (atraso > UMBRAL_ALTO).mean(axis=1),
                         "dif_HM": m_h - m_m,
                         "rechaza": p < ALFA})

sims = {n: simular_cohortes(n) for n in TAMANOS}
print({n: s.shape for n, s in sims.items()})
```

```python
filas = []
for n, s in sims.items():
    pot = s["rechaza"].mean()
    filas.append({
        "n cohorte": n,
        "media atraso": s["media_atraso"].mean(),
        "rango 95% media": f"[{s['media_atraso'].quantile(0.025):.2f}; {s['media_atraso'].quantile(0.975):.2f}]",
        "% alto": s["pct_alto"].mean(),
        "rango 95% % alto": f"[{s['pct_alto'].quantile(0.025):.1f}; {s['pct_alto'].quantile(0.975):.1f}]",
        "dif H−M": s["dif_HM"].mean(),
        "rango 95% dif": f"[{s['dif_HM'].quantile(0.025):.2f}; {s['dif_HM'].quantile(0.975):.2f}]",
        "% cohortes con dif < 0": 100 * (s["dif_HM"] < 0).mean(),
        "potencia Welch %": 100 * pot,
        "EE Monte Carlo potencia": 100 * np.sqrt(pot * (1 - pot) / R),
    })

tabla_mc = pd.DataFrame(filas).set_index("n cohorte")
tabla_mc.round(2)
```

### Interpretación de la simulación Monte Carlo
**Coherencia con la etapa 1.** Promediando las 10.000 cohortes, la simulación reproduce los parámetros de entrada: media de atraso 3,40 (etapa 1: 3,398) y diferencia H−M 1,06–1,07 (etapa 1: 1,068), para todos los tamaños. El % de atraso alto simulado (26,0 %) es 1,5 puntos mayor que el observado en el total (≈ 24,5 %, mezcla de 29,7 % en hombres y 20,5 % en mujeres): la binomial negativa reproduce media y varianza pero engrosa levemente la cola y no genera los atrasos negativos (1,4–2,0 % de los casos reales).

**Convergencia.** Las estimaciones acumuladas se estabilizan desde ≈ 1.000 iteraciones; con R = 10.000 el error Monte Carlo de la potencia es ≤ 0,5 puntos porcentuales, despreciable frente a las diferencias entre tamaños.

**Variabilidad entre cohortes.** Solo por azar, una carrera de 50 titulados puede mostrar entre 14 % y 38 % de atraso alto, y una de 200, entre 20 % y 32,5 %, aun cuando todas provienen del mismo sistema. Diferencias de ese orden entre carreras o años **no** deben interpretarse como cambios reales sin considerar el tamaño de la cohorte.

**Detectar la brecha por sexo.** La diferencia de 1,07 semestres es real a nivel nacional, pero pequeña (d = 0,23). En una cohorte de 200 titulados la prueba de Welch la detecta solo en el 31 % de los casos, y en el 5,9 % de las cohortes la diferencia observada sale invertida (mujeres más atrasadas). Se requieren cerca de 700 titulados para alcanzar 80 % de potencia (68 % con 500 y 94 % con 1.000), lo que concuerda con el cálculo analítico para d = 0,23 (≈ 620).

**Implicancia.** El hallazgo de la etapa 1 es robusto a escala nacional, pero una institución que repita el análisis en su propia cohorte probablemente **no** encontrará diferencias significativas, y eso no contradice el resultado nacional: refleja falta de potencia, no ausencia de efecto. Para análisis institucionales se recomienda agrupar varias cohortes o reportar intervalos de confianza en vez de solo el valor *p*.

## 7. Análisis de robustez
Se evalúa qué conclusiones de la etapa 1 se sostienen al cambiar los datos o los supuestos:
1. **Outliers**: se recalculan los resultados excluyendo los atípicos (Tukey) e inconsistencias marcados en la etapa 1.
2. **Estimadores robustos**: media frente a media recortada y mediana.
3. **Observaciones influyentes (jackknife por institución)**: se excluye una universidad a la vez para ver si alguna determina la diferencia por sexo.
4. **Supuestos**: umbral de "atraso alto" y distribución usada en la simulación Monte Carlo.
5. **Heterogeneidad**: la diferencia por sexo dentro de cada área de conocimiento, para descartar que sea un efecto de composición.

Cada conclusión se clasifica al final como **robusta** o **requiere cautela**.

```python
def resultados_ref(d):
    """Recalcula los resultados principales de la etapa 1 sobre un subconjunto d."""
    h = d.loc[d["sexo"] == "Hombre", "atraso_semestres"].dropna()
    m = d.loc[d["sexo"] == "Mujer",  "atraso_semestres"].dropna()
    sp = np.sqrt(((len(h) - 1) * h.var() + (len(m) - 1) * m.var()) / (len(h) + len(m) - 2))
    welch = stats.ttest_ind(h, m, equal_var=False)
    j = d[d["jornada"].isin(["Diurna", "Vespertina"]) & d["atraso_semestres"].notna()]
    pct = j.groupby("jornada")["atraso_alto"].mean() * 100
    chi2 = stats.chi2_contingency(pd.crosstab(j["jornada"], j["atraso_alto"]), correction=False)
    return {"n": len(h) + len(m),
            "media atraso": d["atraso_semestres"].mean(),
            "dif H−M": h.mean() - m.mean(),
            "t Welch": welch.statistic,
            "d Cohen": (h.mean() - m.mean()) / sp,
            "brecha jornada (pp)": pct["Vespertina"] - pct["Diurna"],
            "V Cramér": np.sqrt(chi2.statistic / len(j)),
            "rho atraso–edad": d["atraso_semestres"].corr(d["edad_titulacion"], method="spearman")}

escenarios = {
    "Datos completos (etapa 1)":      df,
    "Sin atípicos de semestres": df[~df["atipico_semestres"]],
    "Sin atípicos de edad":      df[~df["atipico_edad"]],
    "Sin inconsistentes":        df[~df["incons_antes_formal"]],
    "Sin ninguno de los tres":   df[~(df["atipico_semestres"] | df["atipico_edad"] | df["incons_antes_formal"])],
}
tabla_outliers = pd.DataFrame({nombre: resultados_ref(d) for nombre, d in escenarios.items()}).T
tabla_outliers.round(3)
```

```python
filas = []
for nombre, x in [("Atraso total", df["atraso_semestres"].dropna()),
                  ("Hombres", df.loc[df["sexo"] == "Hombre", "atraso_semestres"].dropna()),
                  ("Mujeres", df.loc[df["sexo"] == "Mujer", "atraso_semestres"].dropna())]:
    filas.append({"grupo": nombre,
                  "media": x.mean(),
                  "media recortada 5%": stats.trim_mean(x, 0.05),
                  "media recortada 10%": stats.trim_mean(x, 0.10),
                  "mediana": x.median()})

tabla_robustos = pd.DataFrame(filas).set_index("grupo")
tabla_robustos.loc["Diferencia H − M"] = tabla_robustos.loc["Hombres"] - tabla_robustos.loc["Mujeres"]
tabla_robustos.round(3)
```

```python
base = df[df["atraso_semestres"].notna()]
agr = base.groupby(["cod_inst", "sexo"])["atraso_semestres"].agg(["sum", "count"]).unstack("sexo", fill_value=0)
S_h, S_m = agr[("sum", "Hombre")], agr[("sum", "Mujer")]
N_h, N_m = agr[("count", "Hombre")], agr[("count", "Mujer")]

# Diferencia H − M dejando fuera cada institución (se resta su aporte a las sumas totales)
dif_sin_i = (S_h.sum() - S_h) / (N_h.sum() - N_h) - (S_m.sum() - S_m) / (N_m.sum() - N_m)
dif_total = S_h.sum() / N_h.sum() - S_m.sum() / N_m.sum()
g_inst = len(dif_sin_i)
ee_jack = np.sqrt((g_inst - 1) / g_inst * ((dif_sin_i - dif_sin_i.mean()) ** 2).sum())

nombres = base.groupby("cod_inst")["nomb_inst"].first()
influencia = pd.DataFrame({"institución": nombres, "titulados": N_h + N_m,
                           "dif H−M sin ella": dif_sin_i, "cambio": dif_sin_i - dif_total})
influencia = influencia.reindex(influencia["cambio"].abs().sort_values(ascending=False).index)

print(f"Instituciones: {g_inst} | dif H−M total = {dif_total:.3f} | EE jackknife = {ee_jack:.3f}")
print(f"Rango al excluir una institución: [{dif_sin_i.min():.3f}; {dif_sin_i.max():.3f}]")
display(influencia.head(5).round(3))

fig, ax = plt.subplots(figsize=(10, 3.8))
orden = dif_sin_i.sort_values()
ax.scatter(range(g_inst), orden.values, s=12, color="0.3")
ax.axhline(dif_total, color="tab:red", linewidth=1.5, label=f"Todas las instituciones = {dif_total:.3f}")
ax.set_xlabel("Institución excluida (ordenadas por resultado)")
ax.set_ylabel("Diferencia H − M (semestres)")
ax.set_title("Jackknife por institución: diferencia de atraso H − M al excluir cada universidad")
ax.legend(fontsize=8)
fig.tight_layout()
fig.savefig(RUTA_FIG / "fig_jackknife_instituciones.png", dpi=300)
plt.show()
```

```python
def potencia_cohorte(generar, n=200, R=R):
    """Potencia de Welch y % atraso alto en R cohortes de n, con el atraso generado por generar(media, de, tamaño)."""
    mujer = rng.random((R, n)) < P_MUJER
    atraso = np.where(mujer, generar(REF_WELCH["media_m"], REF_WELCH["sd_m"], (R, n)),
                             generar(REF_WELCH["media_h"], REF_WELCH["sd_h"], (R, n))).astype(float)
    n_m = mujer.sum(axis=1)
    n_h = n - n_m
    m_m = np.where(mujer, atraso, 0).sum(axis=1) / n_m
    m_h = np.where(~mujer, atraso, 0).sum(axis=1) / n_h
    v_m = np.where(mujer, (atraso - m_m[:, None]) ** 2, 0).sum(axis=1) / (n_m - 1)
    v_h = np.where(~mujer, (atraso - m_h[:, None]) ** 2, 0).sum(axis=1) / (n_h - 1)
    ee2_h, ee2_m = v_h / n_h, v_m / n_m
    t = (m_h - m_m) / np.sqrt(ee2_h + ee2_m)
    gl = (ee2_h + ee2_m) ** 2 / (ee2_h ** 2 / (n_h - 1) + ee2_m ** 2 / (n_m - 1))
    rechaza = 2 * stats.t.sf(np.abs(t), gl) < ALFA
    return 100 * rechaza.mean(), 100 * (atraso > UMBRAL_ALTO).mean()

distribuciones = {
    "Binomial negativa (modelo base)": lambda mu, de, size: rng.negative_binomial(*params_nb(mu, de), size=size),
    "Normal (misma media y DE)":       lambda mu, de, size: rng.normal(mu, de, size=size),
    "Poisson (misma media)":           lambda mu, de, size: rng.poisson(mu, size=size),
}

filas = []
for nombre, generar in distribuciones.items():
    pot, alto = potencia_cohorte(generar)
    filas.append({"distribución del atraso": nombre,
                  "potencia Welch % (n = 200)": pot, "% atraso alto": alto})

tabla_dist = pd.DataFrame(filas).set_index("distribución del atraso")
tabla_dist.round(2)
```

```python
def dif_por_grupo(g):
    h = g.loc[g["sexo"] == "Hombre", "atraso_semestres"]
    m = g.loc[g["sexo"] == "Mujer",  "atraso_semestres"]
    w = stats.ttest_ind(h, m, equal_var=False)
    ic = w.confidence_interval(0.95)
    return pd.Series({"n": len(g), "% hombres": 100 * len(h) / len(g),
                      "atraso medio": g["atraso_semestres"].mean(),
                      "dif H−M": h.mean() - m.mean(), "IC95 inf": ic.low, "IC95 sup": ic.high,
                      "p": w.pvalue})

tabla_area = (base.groupby("area_conocimiento")[["sexo", "atraso_semestres"]]
                  .apply(dif_por_grupo).sort_values("dif H−M"))
dif_ajustada = np.average(tabla_area["dif H−M"], weights=tabla_area["n"])
print(f"Diferencia H − M global: {dif_total:.3f} | promedio ponderado dentro de áreas: {dif_ajustada:.3f}")
display(tabla_area.round(3))

fig, ax = plt.subplots(figsize=(9, 4))
y = np.arange(len(tabla_area))
ax.errorbar(tabla_area["dif H−M"], y,
            xerr=[tabla_area["dif H−M"] - tabla_area["IC95 inf"], tabla_area["IC95 sup"] - tabla_area["dif H−M"]],
            fmt="o", color="black", capsize=3)
ax.set_yticks(y, labels=tabla_area.index)
ax.axvline(0, color="black", linestyle="--", linewidth=1)
ax.axvline(dif_total, color="tab:red", linewidth=1.5, label=f"Global = {dif_total:.2f}")
ax.set_xlabel("Diferencia de atraso H − M (semestres), IC 95 % Welch")
ax.set_title("Brecha por sexo dentro de cada área de conocimiento")
ax.legend(fontsize=8)
fig.tight_layout()
fig.savefig(RUTA_FIG / "fig_brecha_por_area.png", dpi=300)
plt.show()
```

### Síntesis de robustez
**Outliers y estimadores robustos.** Excluir atípicos o inconsistencias nunca cambia el signo ni la significancia de los resultados de la etapa 1, pero sí la magnitud de la brecha por sexo: sin atípicos de semestres baja de 1,07 a 0,58 semestres, con medias recortadas a 0,86–0,90, y la mediana es igual (2) en ambos sexos. La brecha está en la cola: hay más hombres con atrasos muy largos, no un desplazamiento de toda la distribución. Los tamaños de efecto estandarizados (d, V) se mantienen o aumentan.

**Observaciones influyentes.** Excluir cualquiera de las 55 universidades deja la diferencia H−M entre 1,01 y 1,10; la más influyente es la U. Técnica Federico Santa María (sin ella, 1,01). Ninguna institución determina el resultado. Sin embargo, el **EE jackknife por institución (0,111) es 3,5 veces el EE de Welch (0,032)**: los titulados de una misma universidad no son independientes, por lo que los IC de la etapa 1 subestiman la incertidumbre. Con el EE jackknife el IC 95 % es ≈ [0,85; 1,29], más ancho pero aún lejos de 0.

**Supuestos.** La brecha por jornada se mantiene con cualquier umbral de atraso alto y crece con él (5,2 pp con > 3 a 12,1 pp con > 6). En la simulación, la potencia depende de la varianza supuesta y no de la forma (binomial negativa ≈ normal ≈ 33 %; Poisson, que ignora la sobredispersión, 98 %), mientras que el % de atraso alto depende de la forma (normal: 44 %, irreal).

**Heterogeneidad (efecto de composición).** Dentro de cada área de conocimiento la brecha H−M va de −0,07 (Derecho, n.s.) a 1,21 (Agropecuaria), y su promedio ponderado es **0,55 semestres, la mitad de la brecha global**. La otra mitad se explica porque los hombres se concentran en áreas con más atraso (Tecnología: 74 % hombres, atraso medio 4,6; Salud: 26 %, 2,5).

| Conclusión de la etapa 1 | Veredicto | Motivo |
|---|---|---|
| Atraso medio 3,40 semestres; IC [3,37; 3,43] | Robusta (con cautela en la magnitud) | IC validado por bootstrap; la media baja a 2,5–2,8 sin atípicos o con recorte: describe un promedio arrastrado por la cola |
| Atraso típico (mediana) de 2 semestres | Robusta | Igual en todos los escenarios y en ambos sexos |
| Los hombres se atrasan más que las mujeres | Robusta en el signo | Significativa en todos los escenarios, con permutación y sin cada institución |
| La brecha por sexo es de 1,07 semestres | **Requiere cautela** | Mitad composición por área (0,55 dentro de áreas), mitad cola de atípicos (0,58 sin ellos), IC subestimado por agrupamiento institucional |
| Vespertina tiene más atraso alto que diurna | Robusta | Se mantiene con cualquier umbral y sin atípicos; crece con el umbral |

```python
def fila(categoria, parametro, estimado, li=np.nan, ls=np.nan, metodo_ic="", ee=np.nan,
         metodo_ee="", p=np.nan, veredicto="", nota=""):
    return {"categoría": categoria, "parámetro": parametro, "estimado": estimado,
            "IC95 inf": li, "IC95 sup": ls, "método IC": metodo_ic, "EE": ee, "método EE": metodo_ee,
            "p": p, "veredicto": veredicto, "nota": nota}

filas = []

# 1. Parámetros con bootstrap BCa (sección 3)
veredictos = {
    "media atraso":    ("robusta (cautela en magnitud)", "Baja a 2,5–2,8 sin atípicos o con media recortada"),
    "media edad":      ("robusta", "IC t, percentil y BCa coinciden"),
    "dif. atraso H−M": ("requiere cautela", "Signo robusto; magnitud: mitad efecto de área, sensible a atípicos"),
}
for nombre, (theta, reps, jack) in boot.items():
    (li, ls), _, _ = ic_bca(reps, theta, jack)
    es_dif = nombre == "dif. atraso H−M"
    filas.append(fila("Parámetros", nombre, theta, li, ls, "bootstrap BCa (B = 10.000)",
                      ee_jack if es_dif else reps.std(ddof=1),
                      "jackknife por institución" if es_dif else "bootstrap",
                      p1 if es_dif else np.nan, *veredictos[nombre]))

filas.append(fila("Parámetros", "mediana atraso", np.median(x_atraso), veredicto="robusta",
                  nota="Igual a 2 en hombres y mujeres y en todos los escenarios"))
filas.append(fila("Parámetros", "dif. atraso H−M ajustada por área", dif_ajustada,
                  veredicto="requiere cautela", nota="Promedio ponderado de la brecha dentro de 10 áreas"))
filas.append(fila("Parámetros", "brecha % atraso alto Vesp − Diur (pp)", obs2, p=p2,
                  metodo_ee="test de permutación (B = 10.000)", veredicto="robusta",
                  nota="Crece con el umbral: 5,2 pp (> 3) a 12,1 pp (> 6)"))

# 2. Correlaciones de Spearman (sección 5)
for par, r in tabla_corr.iterrows():
    filas.append(fila("Correlaciones", par, r["rho"], r["IC95 inf"], r["IC95 sup"],
                      "bootstrap percentil (B = 10.000)", r["EE boot"], "bootstrap",
                      veredicto="robusta", nota=f"Fuerza {r['fuerza']}"))

# 3. Parámetros del modelo de simulación (sección 6)
for sexo, (r_nb, p_nb) in {"hombres": NB_H, "mujeres": NB_M}.items():
    filas.append(fila("Modelo Monte Carlo", f"BN {sexo}: r", r_nb, nota="Método de momentos con media y DE de la etapa 1"))
    filas.append(fila("Modelo Monte Carlo", f"BN {sexo}: p", p_nb, nota="Método de momentos con media y DE de la etapa 1"))
filas.append(fila("Modelo Monte Carlo", "P(mujer)", P_MUJER, nota="Proporción observada en la etapa 1"))

for n, r in tabla_mc.iterrows():
    filas.append(fila("Potencia Welch", f"cohorte n = {n} (%)", r["potencia Welch %"],
                      ee=r["EE Monte Carlo potencia"], metodo_ee="error Monte Carlo (R = 10.000)",
                      veredicto="requiere cautela" if r["potencia Welch %"] < 80 else "robusta",
                      nota="Bajo el supuesto binomial negativa"))

tabla_validada = pd.DataFrame(filas)
tabla_validada.round(4)
```

```python
# a) Mediana: el atraso es discreto, así que cada remuestra se resume por sus conteos (multinomial)
valores, conteos = np.unique(x_atraso, return_counts=True)
n_x = len(x_atraso)
acum = np.cumsum(rng.multinomial(n_x, conteos / n_x, size=B), axis=1)
reps_mediana = valores[(acum >= (n_x + 1) / 2).argmax(axis=1)]

# b) Brecha H−M ajustada por área: bootstrap estratificado por área × sexo
reps_area = np.zeros(B)
for area, g in base.groupby("area_conocimiento"):
    h_a = g.loc[g["sexo"] == "Hombre", "atraso_semestres"].to_numpy()
    m_a = g.loc[g["sexo"] == "Mujer",  "atraso_semestres"].to_numpy()
    reps_area += len(g) * (boot_media(h_a) - boot_media(m_a))
reps_area /= tabla_area["n"].sum()

# c) Brecha de % atraso alto Vespertina − Diurna: bootstrap estratificado por jornada
alto = 100 * sub2["atraso_alto"].to_numpy(dtype=float)
vesp = (sub2["jornada"] == "Vespertina").to_numpy()
reps_jornada = boot_media(alto[vesp]) - boot_media(alto[~vesp])

# Completar la tabla de resultados validados
nuevos = {"mediana atraso": reps_mediana,
          "dif. atraso H−M ajustada por área": reps_area,
          "brecha % atraso alto Vesp − Diur (pp)": reps_jornada}
for par, reps in nuevos.items():
    i = tabla_validada["parámetro"] == par
    li, ls = ic_percentil(reps)
    tabla_validada.loc[i, ["IC95 inf", "IC95 sup", "EE"]] = [li, ls, reps.std(ddof=1)]
    tabla_validada.loc[i, ["método IC", "método EE"]] = ["bootstrap percentil (B = 10.000)", "bootstrap"]

# Indicar de dónde sale cada p
con_p = tabla_validada["p"].notna()
tabla_validada.loc[con_p, "nota"] += "; p por test de permutación (B = 10.000)"

# Lo que no aplica se muestra vacío
tabla_validada.round(4).fillna("")
```
